```python
# ==========================================================
# Experiment 12
# Deploying a Generative AI Application using Gradio
# ==========================================================

# Install Required Libraries
!pip install -q transformers gradio accelerate sentencepiece

# ----------------------------------------------------------
# Import Libraries
# ----------------------------------------------------------

from transformers import pipeline
import gradio as gr

# ----------------------------------------------------------
# Load GPT-2 Model
# ----------------------------------------------------------

generator = pipeline(
    "text-generation",
    model="gpt2"
)

# ----------------------------------------------------------
# Prediction Function
# ----------------------------------------------------------

def generate_text(prompt):

    result = generator(
        prompt,
        max_new_tokens=60,
        do_sample=True,
        temperature=0.8
    )

    return result[0]["generated_text"]

# ----------------------------------------------------------
# Create Gradio Interface
# ----------------------------------------------------------

demo = gr.Interface(
    fn=generate_text,
    inputs=gr.Textbox(
        lines=3,
        placeholder="Enter your prompt here..."
    ),
    outputs=gr.Textbox(
        lines=8,
        label="Generated Text"
    ),
    title="Generative AI Text Generator",
    description="Generate text using GPT-2 with Hugging Face and Gradio."
)

# ----------------------------------------------------------
# Launch Application
# ----------------------------------------------------------

demo.launch(share=True)
```
