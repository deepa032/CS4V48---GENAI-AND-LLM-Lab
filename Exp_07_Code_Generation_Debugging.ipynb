```python
# ==========================================================
# Experiment 7
# Code Generation and Debugging Using Foundation Models
# ==========================================================

# Install Required Libraries
!pip install -q transformers accelerate sentencepiece

# ----------------------------------------------------------
# Import Library
# ----------------------------------------------------------

from transformers import pipeline

# ----------------------------------------------------------
# Load Code Generation Model
# ----------------------------------------------------------

generator = pipeline(
    "text-generation",
    model="Salesforce/codegen-350M-mono"
)

# ==========================================================
# Part A : Code Generation
# ==========================================================

prompt = """
# Python function to calculate factorial

def factorial(n):
"""

generated = generator(
    prompt,
    max_new_tokens=80,
    do_sample=False
)

print("=" * 60)
print("CODE GENERATION")
print("=" * 60)
print(generated[0]["generated_text"])

# ==========================================================
# Part B : Code Debugging
# ==========================================================

buggy_code = """
# Buggy Python code

def add_numbers(a, b)
    return a + b
"""

debug_prompt = f"""
Fix the following Python code:

{buggy_code}

Correct Code:
"""

corrected = generator(
    debug_prompt,
    max_new_tokens=100,
    do_sample=False
)

print("\n" + "=" * 60)
print("CODE DEBUGGING")
print("=" * 60)
print(corrected[0]["generated_text"])
```
